```python
import numpy as np
```

```python
arr = np.array([10, np.nan, 30, np.nan, 50])

np.isnan(arr)
```

```python
arr = np.array([10, np.inf, 30, -np.inf, 50])

np.isinf(arr)
```

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

np.dot(a, b)
```

```python
arr = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])
np.transpose(arr)
```

```python
arr.T
```

```python
arr = np.array([10, 20, 30, 40, 50])

np.sum(arr)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
```

```python
np.sum(arr,axis=1)
```

```python
np.sum(arr,axis=0)
```

```python
arr = np.array([10, 20, 30, 40, 50])

np.mean(arr)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
np.mean(arr,axis=0)
```

```python
np.mean(arr,axis=1)
```

```python
arr = np.array([10, 20, 30, 40, 50])

np.var(arr)
```
